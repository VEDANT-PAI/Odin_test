```python
## Imports
import pandas as pd
import numpy as np
from pathlib import Path

from sklearn.metrics.pairwise import cosine_similarity
```

```python
## Load cleaned datasets
ROOT = Path("..")

DATASET = ROOT / "datasets" / "dataset_processed"

books = pd.read_csv(DATASET / "books_clean.csv")
ratings = pd.read_csv(DATASET / "ratings_clean.csv")
```

```python
## Dataset overview
print(books.shape)
print(ratings.shape)

ratings.head()
```

```python
## User-Item matrix
user_item = ratings.pivot_table(
    index="user_id",
    columns="book_id",
    values="rating"
)

user_item.head()
```

```python
## Item similarity
item_similarity = cosine_similarity(
    user_item.fillna(0).T
)
```

```python
## Convert to Dataframe
item_similarity = pd.DataFrame(
    item_similarity,
    index=user_item.columns,
    columns=user_item.columns
)
```

```python
def recommend(book_title, n=10):
    # Find matching book
    matches = books[
        books["title"].str.lower() == book_title.lower()
    ]

    if matches.empty:
        print("Book not found.")
        return

    book_id = matches.iloc[0]["book_id"]

    # Get similarity scores
    scores = item_similarity.loc[book_id]

    # Remove the book itself
    scores = scores.drop(book_id)

    # Top N similar books
    top_books = scores.sort_values(ascending=False).head(n)

    recommendations = books[
        books["book_id"].isin(top_books.index)
    ][["title", "authors", "average_rating"]]

    return recommendations
```

```python
recommend("The Hobbit")
```

```python
## Baseline Summary 
- The baseline implementation computes a dense item-item cosine similarity matrix for simplicity. 
- This approach is computationally expensive and memory-intensive on larger datasets. 
- Later notebooks replace this with more scalable nearest-neighbor search and graph-based methods.
```
