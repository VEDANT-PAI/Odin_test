```python
## Imports
import pandas as pd
import numpy as np
from pathlib import Path

pd.set_option("display.max_columns", None)
pd.set_option("display.max_colwidth", 100)
```

```python
## Paths
ROOT = Path("..")

DATASET = ROOT / "datasets" / "dataset_raw"
PROCESSED = ROOT / "datasets" / "dataset_processed"

PROCESSED.mkdir(exist_ok=True)
```

```python
## Load Datasets

books = pd.read_csv(DATASET / "books.csv")
ratings = pd.read_csv(DATASET / "ratings.csv")
tags = pd.read_csv(DATASET / "tags.csv")
book_tags = pd.read_csv(DATASET / "book_tags.csv")
to_read = pd.read_csv(DATASET / "to_read.csv")
```

```python
## Dataset shapes

datasets = {
    "Books": books,
    "Ratings": ratings,
    "Tags": tags,
    "Book Tags": book_tags,
    "To Read": to_read
}

for name, df in datasets.items():
    print(f"{name}: {df.shape}")
```

```python
## Remove Duplicate Rows

print("Duplicates before:", book_tags.duplicated().sum())

book_tags = book_tags.drop_duplicates()

print("Duplicates after :", book_tags.duplicated().sum())
```

```python
## Missing values
books.isnull().sum().sort_values(ascending=False)
```

```python
## Fill missing original titles
books["original_title"] = books["original_title"].fillna(books["title"])
```

```python
## Invalid Publication Year 
books[
    books["original_publication_year"] <= 0
][["title","original_publication_year"]]

### The dataset contains negative publication years for classical works written before the Common Era (BCE). These values are historically meaningful and will be retained.
```

```python
## Invalid Publication Year cleaning
books["original_publication_year"].isna().sum()
```

```python
## Language codes
books["language_code"].value_counts(dropna=False)
```

```python
## Language standardisation
books["language_code"] = books["language_code"].replace({
    "en": "eng",
    "en-US": "eng",
    "en-GB": "eng",
    "en-CA": "eng"
})
```

```python
## Verificarion of Language standardisation
books["language_code"].value_counts(dropna=False)
```

```python
## ISBN
books[["isbn","isbn13"]].isna().sum()
```

```python
## Verify
books.isna().sum()
```

```python
## Save cleaned datasets
books.to_csv(
    PROCESSED / "books_clean.csv",
    index=False
)

ratings.to_csv(
    PROCESSED / "ratings_clean.csv",
    index=False
)

tags.to_csv(
    PROCESSED / "tags_clean.csv",
    index=False
)

book_tags.to_csv(
    PROCESSED / "book_tags_clean.csv",
    index=False
)

to_read.to_csv(
    PROCESSED / "to_read_clean.csv",
    index=False
)
```

```python
# Preprocessing Summary

The following preprocessing steps were performed:

- Removed duplicate rows from `book_tags.csv`.
- Filled missing values in `original_title` using the corresponding `title`.
- Standardized English language codes (`en`, `en-US`, `en-GB`, `en-CA`) to `eng`.
- Retained BCE publication years because they represent historically valid publication dates.
- Left missing values in `isbn`, `isbn13`, `language_code`, and `original_publication_year` unchanged, as they are either identifiers, optional metadata, or too few to justify imputation.

The cleaned datasets were saved for use in the modeling phase.
```
